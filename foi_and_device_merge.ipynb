```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import re
import os
import glob
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import warnings
```

```python
df = pd.read_csv("foi_folder/foitext2025.txt", 
                  sep="|", 
                  nrows=10000,
                  encoding = 'latin1',
                  low_memory=False)

print(df.shape)
print(df.columns.tolist())
print(df.head(5))
```

```python
path = "/projectnb/sparkgrp/dd4g-muckrock-patient-safety/foi_folder/foitext*.txt"
dfs = []

for file in sorted(glob.glob(path)):
    year = re.search(r'\d{4}', os.path.basename(file)).group()
    print(f"Loading {year}...")
    df = pd.read_csv(file, sep="|", dtype=str, encoding='latin-1')
    df["YEAR"] = int(year)
    dfs.append(df)

combined_df = pd.concat(dfs, ignore_index=True)
print(f"\nTotal rows: {combined_df.shape[0]:,}")
print(f"Memory usage: {combined_df.memory_usage(deep=True).sum() / 1e9:.2f} GB")
```

```python
print(f"Total rows: {combined_df.shape[0]:,}")
print(f"Total columns: {combined_df.shape[1]}")
print(f"\nColumns: {combined_df.columns.tolist()}")
print(f"\nYears covered: {sorted(combined_df['YEAR'].unique())}")
print(f"\nNull counts:\n{combined_df.isnull().sum()}")
```

```python
base_path = "/projectnb/sparkgrp/dd4g-muckrock-patient-safety/device_folder/unzipped/"

dfs = []
for file in sorted(glob.glob(base_path + "DEVICE*.txt")):
    year = re.search(r'\d{4}', os.path.basename(file)).group()
    print(f"Loading {year}...")
    df = pd.read_csv(file, sep="|", encoding='latin-1', 
                     on_bad_lines='skip',
                     low_memory=False)  # fixes mixed dtype warning
    df['YEAR'] = int(year)
    dfs.append(df)

device_df = pd.concat(dfs, ignore_index=True)
print(f"\n Total rows: {device_df.shape[0]:,}")
print(f" Columns: {device_df.columns.tolist()}")
```

```python
print(f"Device duplicates: {device_df.duplicated('MDR_REPORT_KEY').sum():,}")

# Keep only first device entry per report
device_dedup = device_df.drop_duplicates(subset='MDR_REPORT_KEY', keep='first')
print(f"Device rows after dedup: {device_dedup.shape[0]:,}")
```

```python
print(f"FOI total rows: {combined_df.shape[0]:,}")
print(f"FOI unique MDR_REPORT_KEY: {combined_df['MDR_REPORT_KEY'].nunique():,}")
print(f"FOI duplicates: {combined_df.duplicated('MDR_REPORT_KEY').sum():,}")

# Check D vs N distribution
print(f"\nTEXT_TYPE_CODE distribution:")
print(combined_df['TEXT_TYPE_CODE'].value_counts())
```

```python
# Merge FOI with Device on MDR_REPORT_KEY
merged_df = pd.merge(
    combined_df,      #  FOI data
    device_df,        #  device catalog
    on='MDR_REPORT_KEY',
    how='left'       # use left as a device could have multiple complaints
)

print(f"FOI rows: {combined_df.shape[0]:,}")
print(f"Device rows: {device_df.shape[0]:,}")
print(f"Merged rows: {merged_df.shape[0]:,}")
print(f"\nAll columns after merge:")
for col in merged_df.columns.tolist():
    print(col)
```

```python
matched = merged_df['BRAND_NAME'].notna().sum() 
print(f"\nMatched rows: {matched:,} ({matched/len(merged_df)*100:.1f}%)")
```

```python
foi_clean = pd.read_csv(
    "/projectnb/sparkgrp/dd4g-muckrock-patient-safety/d1.csv",
    encoding='latin-1',
    low_memory=False
)

print(f"Rows: {foi_clean.shape[0]:,}")
print(f"Columns: {foi_clean.columns.tolist()}")
foi_clean.head()
```

```python
print(foi_clean['FOI_TEXT'].value_counts().head())

print(f"\nTotal rows: {foi_clean.shape[0]:,}")
print(f"Unique MDR_REPORT_KEY: {foi_clean['MDR_REPORT_KEY'].nunique():,}")
print(f"Duplicated keys: {foi_clean.duplicated('MDR_REPORT_KEY').sum():,}")
```

```python
foi_unique = foi_clean.groupby(['MDR_REPORT_KEY', 'YEAR'])['FOI_TEXT'].apply(
    lambda x: ' '.join(x.dropna().astype(str))
).reset_index()

print(f"Before: {foi_clean.shape[0]:,}")
print(f"After concat: {foi_unique.shape[0]:,}")
print(f"Should match unique keys: {foi_clean['MDR_REPORT_KEY'].nunique():,}")
foi_unique.head()
```

```python
print(f"Shape: {foi_unique.shape[0]:,} rows x {foi_unique.shape[1]} cols")
print(f"\nYearly distribution:")
print(foi_unique['YEAR'].value_counts().sort_index())
print(f"\nText length stats:")
foi_unique['TEXT_LENGTH'] = foi_unique['FOI_TEXT'].str.len()
print(foi_unique['TEXT_LENGTH'].describe())
print(f"\nNull FOI_TEXT: {foi_unique['FOI_TEXT'].isna().sum():,}")
print(f"Sample text:\n{foi_unique['FOI_TEXT'].iloc[0][:300]}")
```

```python
print(f"Empty strings: {(foi_unique['FOI_TEXT'] == '').sum():,}")
print(f"'nan' strings: {(foi_unique['FOI_TEXT'] == 'nan').sum():,}")

# Fix text length to int
foi_unique['TEXT_LENGTH'] = foi_unique['FOI_TEXT'].str.len().fillna(0).astype(int)
print(foi_unique['TEXT_LENGTH'].dtype)
```

```python
print(foi_unique.dtypes)
```

```python
foi_unique['MDR_REPORT_KEY'] = foi_unique['MDR_REPORT_KEY'].astype(str)
device_dedup['MDR_REPORT_KEY'] = device_dedup['MDR_REPORT_KEY'].astype(str)

# Verify
print(f"FOI key dtype: {foi_unique['MDR_REPORT_KEY'].dtype}")
print(f"Device key dtype: {device_dedup['MDR_REPORT_KEY'].dtype}")

# Re-merge
merged_df = pd.merge(
    foi_unique,
    device_dedup,
    on='MDR_REPORT_KEY',
    how='left'
)

print(f"\nFOI rows: {foi_unique.shape[0]:,}")
print(f"Merged rows: {merged_df.shape[0]:,}")
```

```python
null_rates = (merged_df.isnull().sum() / len(merged_df) * 100).round(1).sort_values(ascending=False)
print(null_rates.to_string())
```
